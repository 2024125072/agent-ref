```python
from agent_with_context import ask_agent, create_client
from build_context import create_state
```

```python
SCENARIO = [
    (
        "노트북 구매 조건을 기억해. "
        "예산은 150만원 이하이고, "
        "RAM은 16GB 이상이며, "
        "NVIDIA GPU가 반드시 필요해."
    ),

    (
        "search_laptops 도구를 사용해서 "
        "후보를 검색해줘."
    ),

    (
        "추가로 무게는 중요하지 않고 "
        "화면 크기도 상관없어."
    ),

    (
        "지금까지의 조건과 검색 후보를 "
        "바탕으로 하나만 추천해줘."
    ),
]
```

```python
client = create_client()
model = client.models.list().data[0].id
print(f"Using model: {model}")
```

```python
def run_scenario(client, model, state, recent_turns=None, use_summary=False):

    print("=" * 60)
    print(f"scenario: recent_turns={recent_turns}, use_summary={use_summary}")
    print("=" * 60)

    answers = []

    for i, user_text in enumerate(SCENARIO):

        print(f"\nUSER {i+1}:")
        print(user_text)

        # 마지막 질문에서 실제 Context 출력
        debug = (i == len(SCENARIO) - 1)

        outcome = ask_agent(
            client,
            model,
            state,
            user_text,
            recent_turns,
            use_summary,
            debug=debug
        )

        print(f"*" * 60)
        print(f"\nAGENT [{outcome['status']} / {outcome['iterations']} iterations]:")
        print(outcome["content"])

        answers.append(outcome["content"])

    return state, answers
```

```python
state = create_state()
full_state, full_answers = run_scenario(client, model, state, recent_turns=None, use_summary=False)
```
